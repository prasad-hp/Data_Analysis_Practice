```python
import pandas as pd

df_info = pd.DataFrame({
    'EmployeeID': [101, 102, 103, 104],
    'Name': ['Alice Johnson', 'Bob Smith', 'Charlie Brown', 'Diana Prince'],
    'Department': ['Sales', 'HR', 'IT', 'Sales']
})
```

```python
df_salary = pd.DataFrame({
    'EmployeeID': [101, 102, 104, 105],
    'Salary': [60000, 52000, 70000, 48000]
})
```

```python
df_new = pd.DataFrame({
    'EmployeeID': [106, 107],
    'Name': ['Ethan Ray', 'Fiona Dale'],
    'Department': ['Marketing', 'Finance']
})
```

```python
df_info.merge(right=df_salary, left=df_new, left_on='EmployeeID', right_on='EmployeeID', how='outer')
```
