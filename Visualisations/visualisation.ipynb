```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))  # Note: subplots, not Subplot
fig.suptitle('E-commerce Sales Dashboard', fontsize=16, fontweight='bold')

region_sales = df.groupby('region')['total_amount'].sum().sort_values(ascending=False)

axes[0, 0].bar(region_sales.index, region_sales.values, color='skyblue', edgecolor='navy')
axes[0, 0].set_title('Total Sales by Region', fontweight='bold')
axes[0, 0].set_xlabel('Region')
axes[0, 0].set_ylabel('Total Sales ($)')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# Add value labels on bars
for i, v in enumerate(region_sales.values):
    axes[0, 0].text(i, v + max(region_sales.values) * 0.01, f'${v:,.0f}', 
                    ha='center', va='bottom', fontweight='bold')
```
