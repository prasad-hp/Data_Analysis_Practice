```python

data = {
    'Date': ['2024-01-01', '2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-03'],
    'Region': ['North', 'South', 'East', 'North', 'South', 'East'],
    'Product': ['Apples', 'Apples', 'Bananas', 'Bananas', 'Apples', 'Apples'],
    'Units Sold': [100, 150, 200, 120, 180, 160],
    'Sales Amount': [1000, 1500, 2200, 1300, 1700, 1550]
}

sales_df = pd.DataFrame(data)
```

```python
sales_df
```

```python
sales_df.pivot_table(index='Product', aggfunc='sum')
```

```python
sales_df.pivot_table(values='Sales Amount', index='Region', columns='Product', aggfunc='mean')
```

Total sales by date and region

```python
sales_df.pivot_table(index='Region', columns='Date', values='Sales Amount', aggfunc='sum')
```

```python
sales_df.pivot_table(values='Sales Amount', index='Region', aggfunc='count')
```
