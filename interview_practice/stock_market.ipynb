## Stock Market Analysis

```python
import yfinance as yf
```

```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
%matplotlib inline
```

```python
data = yf.Tickers(['META', "AAPL", 'AMZN', 'NVDA', 'NFLX', 'GOOG'])
```

```python
data
```

```python
ticks = ['META', "AAPL", 'AMZN', 'NVDA', 'NFLX', 'GOOG']
```

```python
stock_data = yf.download(ticks, period='10y')
```

```python
stock_data.head()
```

```python
stock_data.to_csv('stock_data_10y.csv')
```

```python
stock_data.drop(columns=['High', 'Open'], inplace=True)
```

```python
stock_data.head()
```

```python
stock_data.dtypes
```

```python
stock_data['Date'] = stock_data.index
```

```python
stock_data.dtypes
```

```python
fig, ax = plt.subplots(figsize=(15, 5))
sns.lineplot(stock_data['Close'])
plt.title("Price Trend Over Last 10 Years")
plt.ylabel("Stock Price")
plt.grid(visible=True)
```

```python
fig, ax = plt.subplots(figsize=(15, 8))
sns.lineplot(stock_data['Close']['NVDA'])
plt.grid()
```

```python
stock_data['MA20_NVDA'] = stock_data['Close']['NVDA'].rolling(window=20).mean()
stock_data['MA50_NVDA'] = stock_data['Close']['NVDA'].rolling(50).mean()
```

```python
stock_data.drop(columns=['Date', 'MA20'], inplace=True)
```

```python
fig, ax = plt.subplots(figsize=(15,8))
sns.lineplot([stock_data['Close']['NVDA']])
sns.lineplot(stock_data['MA20_NVDA'])
sns.lineplot(stock_data['MA50_NVDA'])
plt.grid()
```

```python
stock_weekly = stock_data['Close'].resample('W').last()
```
