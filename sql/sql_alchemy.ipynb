```python
from sqlalchemy import create_engine, text
import pandas as pd
```

```python
engine = create_engine("postgresql+psycopg2://postgres:gowda@localhost:5432/Practice")
```

```python
with engine.connect() as connection:
    result = connection.execute(text('SELECT * FROM employee'))
    df = pd.DataFrame(result)
```

```python
df.groupby('gender')['salary'].mean()
```

```python
df
```

```python
df['rank_min'] = df['salary'].rank(method='min', ascending=False)
```

```python
df['rank_avg'] = df['salary'].rank(method='average', ascending=False)
```

```python
df['rank_dense'] = df['salary'].rank(method='dense', ascending=False)
```

```python
df['rank_max'] = df['salary'].rank(method='max', ascending=False)
```

```python
df['rank'] = df['salary'].rank(method='first', ascending=False)
```

```python
df.sort_values(by='rank_min')
```
